```python

# -*- coding: utf-8 -*-
"""
Create targeted-pool balance forecasts and a targeted Weights sheet
from '20260109_CLO_NLIC_DerivativesDeskCashflowProjection.xlsx'.

What it does
------------
1) Read the Weights sheet and sum weights across TargetedPools per CUSIP.
2) For each Scenario sheet:
   - Multiply every CUSIP column by its summed targeted weight.
   - Date column is preserved unchanged.
   - Columns with no targeted weight become 0 (so results are "TargetedPools only").
3) Recreate the Weights tab with:
   - Only TargetedPools retaining weight (non-targeted columns set to 0).
   - TargetedPools re-normalized to sum to 1 (100%) per CUSIP when exposure exists.
   - For CUSIPs without any targeted exposure: leave all zeros (including 'All'=0).
4) Preserve workbook structure: same sheets; same rows/columns/order.

Notes
-----
- Handles stray carriage returns or whitespace in pool names.
- If any pool in TargetedPools is not found among columns, it is ignored (warning printed).
- If a column named 'All' exists on the Weights sheet, it is recomputed as 1 where
  targeted exposure exists, else 0 — while keeping the column position.
"""

from pathlib import Path
import pandas as pd
import numpy as np

# ----------------------------------------------------------
# Configuration
# ----------------------------------------------------------
INPUT_FILE  = Path("20260109_CLO_NLIC_DerivativesDeskCashflowProjection.xlsx")
OUTPUT_FILE = INPUT_FILE.with_name(INPUT_FILE.stem + "_TargetedPools.xlsx")

# Pools to target (trimmed to remove any stray \r or spaces)
TargetedPools_raw = ['13015\r', '13035\r', '13036']
TargetedPools = [str(p).strip() for p in TargetedPools_raw]

# We'll treat every sheet whose name starts with this prefix as a scenario sheet
SCENARIO_SHEET_PREFIX = "Scenario"
WEIGHTS_SHEET_NAME = "Weights"

# ----------------------------------------------------------
# Helpers
# ----------------------------------------------------------
def _first_col_name(df: pd.DataFrame) -> str:
    """Return the leftmost column name (used as the CUSIP id column on Weights)."""
    return str(df.columns[0])

def _numeric_cols(df: pd.DataFrame, exclude: list) -> list:
    """All columns except those in `exclude`, preserving order."""
    return [c for c in df.columns if c not in exclude]

def _recompute_all_column(df: pd.DataFrame, cusip_col: str, all_col: str, targeted_sum: pd.Series) -> pd.DataFrame:
    """Set 'All' column to 1 for rows with targeted exposure, else 0."""
    out = df.copy()
    out[all_col] = (targeted_sum > 0).astype(float)
    return out

# ----------------------------------------------------------
# Load workbook
# ----------------------------------------------------------
xls = pd.ExcelFile(INPUT_FILE, engine="openpyxl")

# -------------------------
# Read and prep Weights
# -------------------------
weights = pd.read_excel(xls, sheet_name=WEIGHTS_SHEET_NAME, engine="openpyxl")
weights.columns = [str(c).strip() for c in weights.columns]  # strip col labels

cusip_col = _first_col_name(weights)  # e.g., 'CUSIP'
# Ensure string CUSIP labels (preserve 'All' row etc.)
weights[cusip_col] = weights[cusip_col].astype(str).str.strip()

# Identify numeric pool columns (everything except the leftmost key column)
pool_cols = _numeric_cols(weights, exclude=[cusip_col])

# Coerce numeric values; replace NaNs with 0
weights[pool_cols] = (
    weights[pool_cols]
    .apply(pd.to_numeric, errors="coerce")
    .fillna(0.0)
)

# Normalize TargetedPools list to those actually present in columns
present_pools = [p for p in TargetedPools if p in pool_cols]
missing_pools = [p for p in TargetedPools if p not in pool_cols]
if missing_pools:
    print(f"[WARN] These TargetedPools are not in the Weights columns and will be ignored: {missing_pools}")

# Compute the per-CUSIP sum of targeted weights
if present_pools:
    targeted_sum = weights[present_pools].sum(axis=1)
else:
    # Edge case: none of the targeted pools present → everything zeros out
    targeted_sum = pd.Series(0.0, index=weights.index)

# Build a lookup: CUSIP -> sum of targeted weights (used to scale Scenario balances)
cusip_to_targeted_weight = pd.Series(targeted_sum.values, index=weights[cusip_col]).astype(float)

# -------------------------
# Build targeted Weights sheet
# -------------------------
weights_targeted = weights.copy()

# Zero out all pools first
weights_targeted[pool_cols] = 0.0

# For rows with positive targeted exposure, re-normalize only the targeted pools to sum to 1
# new_weight = original_weight / sum_targeted (others remain 0)
with np.errstate(divide="ignore", invalid="ignore"):
    denom = targeted_sum.replace(0, np.nan)  # to avoid divide by zero
    normalized_targeted = weights[present_pools].div(denom, axis=0)
    normalized_targeted = normalized_targeted.fillna(0.0)

# Set the targeted pool columns to the normalized values
for p in present_pools:
    weights_targeted[p] = normalized_targeted[p].astype(float)

# If there is a column called 'All', recompute as 1 where exposure exists else 0.
# Keep it exactly where it was in the column order.
all_col_name = None
for c in pool_cols:
    if c.lower() == "all":
        all_col_name = c
        break

if all_col_name is not None:
    weights_targeted = _recompute_all_column(weights_targeted, cusip_col, all_col_name, targeted_sum)

# -------------------------
# Process Scenario sheets
# -------------------------
scenario_sheets = [s for s in xls.sheet_names if s.startswith(SCENARIO_SHEET_PREFIX)]
other_sheets    = [s for s in xls.sheet_names if s not in scenario_sheets + [WEIGHTS_SHEET_NAME]]

writer = pd.ExcelWriter(OUTPUT_FILE, engine="openpyxl")

# 1) Write the targeted Weights sheet first (preserve original column order exactly)
weights_targeted.to_excel(writer, sheet_name=WEIGHTS_SHEET_NAME, index=False)

# 2) For each Scenario sheet: scale balances by CUSIP targeted weight; Date column unchanged
for sname in scenario_sheets:
    df = pd.read_excel(xls, sheet_name=sname, engine="openpyxl")
    if df.empty:
        df.to_excel(writer, sheet_name=sname, index=False)
        continue

    # Preserve column order
    cols = list(df.columns)
    # The first column is assumed to be the Date column
    date_col = cols[0]
    balance_cols = cols[1:]  # everything else are identifiers (CUSIPs, tranche IDs, etc.)

    # Build a multiplier vector aligned to the scenario columns
    # Any column not found in the Weights mapping gets weight 0
    multipliers = pd.Series(
        [cusip_to_targeted_weight.get(str(c), 0.0) for c in balance_cols],
        index=balance_cols,
        dtype="float64"
    )

    # Scale balances
    out = df.copy()
    # Only multiply numeric columns and leave non-numeric (if any) intact
    # But here we assume scenario balance columns are numeric
    out[balance_cols] = df[balance_cols].apply(pd.to_numeric, errors="coerce").fillna(0.0)
    out[balance_cols] = (out[balance_cols] * multipliers.values)

    # Write back
    out.to_excel(writer, sheet_name=sname, index=False)

# 3) Copy all other sheets unchanged (structure preserved)
for sname in other_sheets:
    df = pd.read_excel(xls, sheet_name=sname, engine="openpyxl")
    df.to_excel(writer, sheet_name=sname, index=False)

writer.close()

print(f"Done. Wrote: {OUTPUT_FILE.name}")
```
